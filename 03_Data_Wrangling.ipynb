```python
import pandas as pd
import numpy as np

# Course dataset. To use your own file from notebook 01, swap to: pd.read_csv('my_dataset_part_1.csv')
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
df.head()
```

```python
print(df.shape)
print(df.isnull().sum() / len(df) * 100)
print(df.dtypes)
```

## Categories

```python
print(df['LaunchSite'].value_counts(), '\n')
print(df['Orbit'].value_counts(), '\n')
landing_outcomes = df['Outcome'].value_counts()
print(landing_outcomes)
```

## Build the target
`True ...` outcomes mean the first stage landed (Class = 1). `False ...` and `None ...` mean it did not (Class = 0).

```python
df['Class'] = df['Outcome'].str.startswith('True').astype(int)
print(df['Class'].value_counts())
print("overall success rate:", round(df['Class'].mean(), 3))
print("baseline accuracy (always predict landed):", round(max(df['Class'].mean(), 1 - df['Class'].mean()), 3))
```
